```python
# 导入必要的库
import requests
from PIL import Image
from io import BytesIO
import matplotlib.pyplot as plt
import os

# 创建输出目录
os.makedirs("output", exist_ok=True)
```

## 2. 图像生成

Pollinations.AI提供了强大的图像生成功能，支持多种模型，如Stable Diffusion和SDXL。下面我们将定义一个函数来使用Pollinations.AI API生成图像。

```python
def generate_image(prompt, width=512, height=512, model="stable-diffusion"):
    """
    使用Pollinations.AI API生成图像
    
    参数:
        prompt (str): 描述要生成的图像的文本提示
        width (int): 图像宽度
        height (int): 图像高度
        model (str): 要使用的模型，可选值包括"stable-diffusion"和"sdxl"
        
    返回:
        PIL.Image: 生成的图像对象
    """
    # Pollinations.AI API端点
    url = "https://pollinations.ai/api/generate/image"
    
    # 请求参数
    params = {
        "prompt": prompt,
        "width": width,
        "height": height,
        "model": model
    }
    
    print(f"正在生成图像，提示词: '{prompt}'")
    
    # 发送请求
    response = requests.post(url, json=params)
    
    # 检查响应状态
    if response.status_code == 200:
        # 从响应中获取图像URL
        image_url = response.json().get("imageUrl")
        print(f"图像生成成功，URL: {image_url}")
        
        # 下载图像
        image_response = requests.get(image_url)
        image = Image.open(BytesIO(image_response.content))
        
        return image
    else:
        print(f"错误: {response.status_code} - {response.text}")
        return None

def display_image(image, prompt=""):
    """
    显示生成的图像
    
    参数:
        image (PIL.Image): 要显示的图像
        prompt (str): 用于生成图像的提示词
    """
    if image:
        plt.figure(figsize=(10, 10))
        plt.imshow(image)
        plt.axis('off')
        if prompt:
            plt.title(prompt)
        plt.show()
    else:
        print("没有图像可显示")
```

### 2.1 基本图像生成

让我们生成第一张图像。我们将创建一个"中国传统山水画风格的机器人"。

```python
prompt = "一只在中国传统山水画风格中的机器人"
image = generate_image(prompt)
display_image(image, prompt)
```

### 2.2 使用SDXL模型

SDXL是一个更高质量的模型，可以生成更详细的图像。让我们尝试使用SDXL生成一个赛博朋克风格的城市天际线。

```python
prompt = "未来城市天际线，赛博朋克风格，夜晚，霓虹灯"
image = generate_image(prompt, width=768, height=512, model="sdxl")
display_image(image, prompt)
```

## 3. 提示工程（Prompt Engineering）

提示工程是指设计和优化输入提示以获得更好的AI生成结果的过程。下面我们将探索不同类型的提示如何影响生成的图像。

### 3.1 基本提示与详细提示的比较

让我们比较一个基本提示和一个详细提示的效果。

```python
# 基本提示
basic_prompt = "一只猫"
basic_image = generate_image(basic_prompt)
display_image(basic_image, basic_prompt)
```

```python
# 详细提示
detailed_prompt = "一只橙色的猫，坐在窗台上，阳光照射，写实风格，高细节"
detailed_image = generate_image(detailed_prompt)
display_image(detailed_image, detailed_prompt)
```

### 3.2 风格转换

我们可以通过在提示中指定不同的艺术风格来生成不同风格的图像。

```python
styles = ["油画风格", "水彩画风格", "素描风格", "像素艺术风格"]
base_prompt = "一只猫坐在窗台上"

for style in styles:
    prompt = f"{base_prompt}，{style}"
    image = generate_image(prompt)
    display_image(image, prompt)
```

### 3.3 概念混合

我们可以混合不同的概念来创建有趣的图像。

```python
blend_prompt = "半机械半有机的花朵，在科幻实验室中绽放"
blend_image = generate_image(blend_prompt)
display_image(blend_image, blend_prompt)
```

## 4. 文本生成

Pollinations.AI还支持文本生成。下面我们将定义一个函数来使用Pollinations.AI API生成文本。

```python
def generate_text(prompt, max_tokens=100):
    """
    使用Pollinations.AI API生成文本
    
    参数:
        prompt (str): 文本提示
        max_tokens (int): 生成的最大标记数
        
    返回:
        str: 生成的文本
    """
    url = "https://pollinations.ai/api/generate/text"
    
    params = {
        "prompt": prompt,
        "max_tokens": max_tokens
    }
    
    print(f"正在生成文本，提示词: '{prompt}'")
    
    response = requests.post(url, json=params)
    
    if response.status_code == 200:
        text = response.json().get("text")
        print("文本生成成功")
        return text
    else:
        print(f"错误: {response.status_code} - {response.text}")
        return None
```

### 4.1 生成解释性文本

让我们生成一段关于人工智能如何改变教育的文本。

```python
prompt = "解释人工智能如何改变教育："
generated_text = generate_text(prompt, max_tokens=200)
print(generated_text)
```

### 4.2 生成创意内容

我们也可以生成创意内容，如故事或诗歌。

```python
prompt = "写一首关于人工智能和人类合作的短诗："
generated_poem = generate_text(prompt, max_tokens=150)
print(generated_poem)
```

## 5. 实际应用案例

下面我们将探索一些生成式AI的实际应用案例。

### 5.1 教育应用：概念可视化

生成式AI可以帮助将抽象概念转化为可视化图像，使学习更加直观。

```python
concepts = [
    "神经网络的工作原理，概念图",
    "机器学习算法如何从数据中学习，教育插图",
    "人工智能与人类协作，未来教育场景"
]

for concept in concepts:
    image = generate_image(concept, model="sdxl")
    display_image(image, concept)
```

### 5.2 创意应用：故事插图

生成式AI可以为故事创作配图，激发创意写作。

```python
# 首先生成一个短故事
story_prompt = "写一个关于一个年轻发明家和他的机器人助手的短故事："
story = generate_text(story_prompt, max_tokens=300)
print(story)

# 然后为故事生成插图
illustration_prompt = "年轻发明家和他的机器人助手在实验室工作，温馨场景，插图风格"
illustration = generate_image(illustration_prompt, model="sdxl")
display_image(illustration, "故事插图")
```
